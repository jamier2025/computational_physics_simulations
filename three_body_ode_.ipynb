```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

# Normalized Gravitational Constant
G = 1.0

def three_body_ode(t, y, m1, m2, m3):
    # Unpack positions (r) and velocities (v) for 3 bodies in 2D
    r1, r2, r3 = y[0:2], y[2:4], y[4:6]
    v1, v2, v3 = y[6:8], y[8:10], y[10:12]

    # Relative position vectors and distances
    r12 = np.linalg.norm(r2 - r1)
    r13 = np.linalg.norm(r3 - r1)
    r23 = np.linalg.norm(r3 - r2)

    # Gravitational acceleration vectors: a = G * m * r_hat / |r|^2
    a1 = G * m2 * (r2 - r1) / r12**3 + G * m3 * (r3 - r1) / r13**3
    a2 = G * m1 * (r1 - r2) / r12**3 + G * m3 * (r3 - r2) / r23**3
    a3 = G * m1 * (r1 - r3) / r13**3 + G * m2 * (r2 - r3) / r23**3

    return [*v1, *v2, *v3, *a1, *a2, *a3]

# Equal mass bodies
m1, m2, m3 = 1.0, 1.0, 1.0

# Initial Positions (2D coordinates)
r1_0 = np.array([-1.0, 0.0])
r2_0 = np.array([1.0, 0.0])
r3_0 = np.array([0.0, 0.0])

# Initial Velocities (Stable Figure-8 initial state)
v1_0 = np.array([0.347111, 0.532728])
v2_0 = np.array([0.347111, 0.532728])
v3_0 = np.array([-2 * 0.347111, -2 * 0.532728])

y0 = [*r1_0, *r2_0, *r3_0, *v1_0, *v2_0, *v3_0]
t_span = (0, 10)
t_eval = np.linspace(0, 10, 2000)

# High-accuracy Runge-Kutta 8th-order integration
sol = solve_ivp(three_body_ode, t_span, y0, args=(m1, m2, m3), t_eval=t_eval, method='DOP853', rtol=1e-9)

# Trajectory Plot
plt.figure(figsize=(7, 7))
plt.plot(sol.y[0], sol.y[1], 'cyan', label='Body 1', lw=1.2)
plt.plot(sol.y[2], sol.y[3], 'magenta', label='Body 2', lw=1.2)
plt.plot(sol.y[4], sol.y[5], 'yellow', label='Body 3', lw=1.2)

plt.xlabel('X Coordinate')
plt.ylabel('Y Coordinate')
plt.title('Three-Body Gravitational Problem (SciPy DOP853 ODE Solver)')
plt.legend()
plt.axis('equal')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
