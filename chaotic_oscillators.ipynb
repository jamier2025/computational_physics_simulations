```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

# Damped Driven Pendulum ODE: d^2(theta)/dt^2 + gamma*d(theta)/dt + sin(theta) = A*cos(omega*t)
def pendulum_ode(t, y, gamma, A, omega):
    theta, omega_val = y
    dtheta_dt = omega_val
    domega_dt = -gamma * omega_val - np.sin(theta) + A * np.cos(omega * t)
    return [dtheta_dt, domega_dt]

# Parameters for chaotic phase space motion
gamma, A, omega = 0.2, 1.2, 0.666
t_span = (0, 100)
t_eval = np.linspace(0, 100, 5000)
y0 = [0.2, 0.0]  # Initial angle (rad) and angular velocity (rad/s)

sol = solve_ivp(pendulum_ode, t_span, y0, args=(gamma, A, omega), t_eval=t_eval, method='RK45')

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# Time series
ax1.plot(sol.t, sol.y[0], 'b-', lw=0.8)
ax1.set_xlabel('Time t')
ax1.set_ylabel('Angle θ (rad)')
ax1.set_title('Non-Linear Chaotic Pendulum Time Series')

# Phase Space Trajectory (ω vs θ)
ax2.plot(sol.y[0], sol.y[1], 'r-', lw=0.5)
ax2.set_xlabel('Angle θ (rad)')
ax2.set_ylabel('Angular Velocity ω (rad/s)')
ax2.set_title('Phase Space Trajectory (ω vs θ)')

plt.tight_layout()
plt.show()
```
