```python
import numpy as np
import matplotlib.pyplot as plt

# Monte Carlo Compton Scattering Simulation
N_photons = 10000
E_initial = 511.0  # Initial photon energy in keV (rest mass of electron)

# Sample isotropic scattering angles theta in [0, pi]
angles = np.arccos(1 - 2 * np.random.random(N_photons))

# Relativistic Compton Scattering equation: E' = E / (1 + (E/m_e c^2)*(1 - cos(theta)))
E_scattered = E_initial / (1 + (E_initial / 511.0) * (1 - np.cos(angles)))

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# Angular distribution plot
ax1.hist(np.degrees(angles), bins=50, color='cyan', edgecolor='black', alpha=0.7)
ax1.set_xlabel('Scattering Angle θ (degrees)')
ax1.set_ylabel('Photon Count')
ax1.set_title('Monte Carlo Angular Distribution')

# Post-scattering energy distribution plot
ax2.hist(E_scattered, bins=50, color='orange', edgecolor='black', alpha=0.7)
ax2.set_xlabel('Scattered Photon Energy E\' (keV)')
ax2.set_ylabel('Photon Count')
ax2.set_title('Scattered Photon Energy Spectrum')

plt.tight_layout()
plt.show()
```
