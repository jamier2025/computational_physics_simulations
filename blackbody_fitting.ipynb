```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

h, c, k = 6.626e-34, 3.0e8, 1.38e-23

def planck(freq, T):
    return (2 * h * freq**3 / c**2) / (np.exp((h * freq) / (k * T)) - 1)

frequencies = np.linspace(1e13, 1e15, 100)
noisy_data = planck(frequencies, 5800) * (1 + 0.05 * np.random.normal(size=100))
popt, _ = curve_fit(planck, frequencies, noisy_data, p0=[5000])

plt.plot(frequencies, noisy_data, 'r.', label='Observed Data')
plt.plot(frequencies, planck(frequencies, *popt), 'b-', label=f'Fit T = {popt[0]:.1f}K')
plt.xlabel('Frequency (Hz)')
plt.ylabel('Spectral Radiance')
plt.legend()
plt.title('Planck Blackbody Curve Fitting')
plt.show()
```
